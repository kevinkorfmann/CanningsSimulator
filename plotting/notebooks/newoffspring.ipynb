```r
get_schweinsber_param<-function(alpha_x=2,p0_x=0,s=0,optimize_alpha=T){
  results=list()
  DOABLE=T
  if(optimize_alpha){
    p0_y=p0_x
    truc=(1+s)*((alpha_x+1)/(alpha_x) )
    alpha_y=1/(truc-1)
    if(alpha_y<1|alpha_y>2){
      DOABLE=F
    }
  }else{
    alpha_y=alpha_x
    truc=((1+s)^alpha_x)*(1-p0_x)
    p0_y=(1 - truc)
    if(p0_y<0|p0_y>1){
      DOABLE=F
    }
  }
 
  if(DOABLE){
  results$alpha_y=alpha_y
  results$p0_y=p0_y
  }
  return(results)
}
```
