```python
def plot_bar(df, ax, title="", legend=True):
    df["α"] = df["α"].astype("str")
    df["n"] = df["n"].astype("str")
    sns.barplot(data=df, x="n", y="fixation_probability", hue="α",
    hue_order=["2.0","1.9", "1.7", "1.5", "1.3"],
    palette=["black","#03045e", "#0077b6", "#00b4d8", "#90e0ef"],
    edgecolor='black', alpha=0.90, ax=ax).set_title(title)

    #sns.lineplot(data=df, x="n", y="fixation_probability", hue="α",
    #hue_order=["2.0","1.9", "1.7", "1.5", "1.3"],
    #palette=["black","#03045e", "#0077b6", "#00b4d8", "#90e0ef"],
    #alpha=0.90, ax=ax).set_title(title)

    ax.legend_.remove()
    ax.patch.set_edgecolor('black')  
    ax.patch.set_linewidth('2') 
    ax.set(xlabel='', ylabel='Fixation probability')
    ax.tick_params(bottom=True, top=False, left=True, right=True)
    ax.grid(True)

    if legend:
        ax.set(xlabel='Population size', ylabel='Fixation probability')
        box = ax.get_position()
        h, l = ax.get_legend_handles_labels()
        ax.legend(handles=h, loc='upper center',labels=["λ=1.0",
        "α=1.9","α=1.7", "α=1.5", "α=1.3"],
        bbox_to_anchor=(0.5, legend_pos), fancybox=True, shadow=True, ncol=ncol)



sns.set(font_scale = 1.2, style="whitegrid")
font_scale=1
legend_pos=-0.20
ncol = 5

fig, axes = plt.subplots(2, 3, figsize=(14, 8), sharey=True)

ax = axes[0][0]
title = "A) fecundity selection 0.1"
df = pd.read_csv("fig_2_fecund_0.1_schweinsberg_fixation_probabilities.csv")
df2 = pd.read_csv("fig_2_fecund_0.1_poisson_fixation_probabilities.csv")
df = pd.concat([df, df2]).reset_index()
plot_bar(df, ax, title, legend=False)

ax = axes[0][1]
title = "B) fecundity selection 0.05"
df = pd.read_csv("fig_2_fecund_0.05_schweinsberg_fixation_probabilities.csv")
df2 = pd.read_csv("fig_2_fecund_0.05_poisson_fixation_probabilities.csv")
df = pd.concat([df, df2]).reset_index()
plot_bar(df, ax, title, legend=False)

ax = axes[0][2]
title = "C) fecundity selection 0.01"
df = pd.read_csv("fig_2_fecund_0.01_schweinsberg_fixation_probabilities.csv")
df2 = pd.read_csv("fig_2_fecund_0.01_poisson_fixation_probabilities.csv")
df = pd.concat([df, df2]).reset_index()
plot_bar(df, ax, title, legend=False)

ax = axes[1][0]
title = "D) viability selection 0.1"
df = pd.read_csv("fig_2_viable_0.1_schweinsberg_fixation_probabilities.csv")
df2 = pd.read_csv("fig_2_viable_0.1_poisson_fixation_probabilities.csv")
df = pd.concat([df, df2]).reset_index()
plot_bar(df, ax, title, legend=False)
ax.set(xlabel='Population size', ylabel='Fixation probability')


ax = axes[1][1]
title = "E) viability selection 0.05"
df = pd.read_csv("fig_2_viable_0.05_schweinsberg_fixation_probabilities.csv")
df2 = pd.read_csv("fig_2_viable_0.05_poisson_fixation_probabilities.csv")
df = pd.concat([df, df2]).reset_index()
plot_bar(df, ax, title)


ax = axes[1][2]
title = "F) viability selection 0.01"
df = pd.read_csv("fig_2_viable_0.01_schweinsberg_fixation_probabilities.csv")
df2 = pd.read_csv("fig_2_viable_0.01_poisson_fixation_probabilities.csv")
df = pd.concat([df, df2]).reset_index()
plot_bar(df, ax, title=title, legend=False)
ax.set(xlabel='Population size', ylabel='Fixation probability')


#plt.tight_layout()
```

```python
import os
from copy import deepcopy
import numpy as np

files = os.listdir("./all_simulations/")
files = ["./all_simulations/" + file for file in files]
# filter the times
df_time = [pd.read_csv(file) for file in files if "time" in file]
df_time = pd.concat(df_time).reset_index()
df = df_time[df_time["population_size_model"] == "constant"]
```

```python
def pos_to_char(pos):
    return chr(pos + 65)
```

```python
df = df[df["population_size_model"] == "constant"]

#df_viability_constant = df[df["selection_type"] == "viability_constant"].reset_index()
#df_fecundity_constant = df[df["selection_type"] == "fecundity_constant"].reset_index()
#df_viability_sinus = df[df["selection_type"] == "viability_sinus"].reset_index()
#df_fecundity_sinus = df[df["selection_type"] == "fecundity_sinus"].reset_index()
```

```python
fecundity_constant = df[df["selection_type"] == "fecundity_constant"]
viability_constant = df[df["selection_type"] == "viability_constant"]

selection_coefficients = [0, 0.01, 0.1]
population_sizes = [500, 1000, 5000]

difference_frame = []

for population_size in population_sizes:

    for selection_coefficient in selection_coefficients:

        constant = pd.concat([fecundity_constant, viability_constant]).reset_index()
        constant = constant[constant["population_size"] == population_size]
        constant = constant[constant["selection_coefficient"] == selection_coefficient]

        for alpha in [1.1, 1.3, 1.5, 1.7, 1.9, 2.0]:
            alpha_frame = constant[constant["parameter"] == alpha]
            fc = alpha_frame[alpha_frame["selection_type"] == "fecundity_constant"]
            vc = alpha_frame[alpha_frame["selection_type"] == "viability_constant"]
            diff = fc.fixation_time.item() - vc.fixation_time.item()
            diff_std = np.mean([fc.standard_deviation.item(), vc.standard_deviation.item()])

            #print(alpha, diff, population_size, selection_coefficient)

            difference_frame.append([alpha, diff, population_size, selection_coefficient, diff_std])
            
difference_frame = pd.DataFrame(difference_frame, columns=["alpha", "fecundity-viability", "population_size", "selection_coefficient", "standard_deviation"])
```

```python
colors = ["#03045e", "#0077b6", "#00b4d8"]
#colors = ["#03045e", "#0077b6"]



sns.set(font_scale = 1.3, style="whitegrid")
fig, axs = plt.subplots(1, 3, figsize=(16, 5), sharey=True)

for i, population_size in enumerate(population_sizes):

    difference_frame_population__size = difference_frame[difference_frame["population_size"] == population_size]
    ax = axs[i]
    sns.lineplot(data=difference_frame_population__size, x="alpha", y="fecundity-viability", hue="selection_coefficient", ax=ax, palette=colors, linewidth=2, legend=False).set_title(pos_to_char(i) + ") fecundity-viability selection (N=" + str(population_size) + ")")
    if i != 1:
        sns.scatterplot(data=difference_frame_population__size, x="alpha", y="fecundity-viability", hue="selection_coefficient", ax=ax, palette=colors, legend=False)
    else:
        sns.scatterplot(data=difference_frame_population__size, x="alpha", y="fecundity-viability", hue="selection_coefficient", ax=ax, palette=colors)

    for j, selection_coefficient in enumerate(selection_coefficients):
        std_frame = difference_frame_population__size[difference_frame_population__size["selection_coefficient"] == selection_coefficient]
        ax.fill_between(std_frame["alpha"], y1=std_frame["fecundity-viability"] - std_frame["standard_deviation"], y2=std_frame["fecundity-viability"] + std_frame["standard_deviation"], alpha=.15, color= colors[j])

    ax.set(xlabel='α', ylabel='mean fixation time')
    ax.grid(True)
    ax.tick_params(bottom=True, top=False, left=True, right=True)
    ax.patch.set_edgecolor('black')  
    ax.patch.set_linewidth('2') 


legend_pos=-0.25    
ncol=3
ax=axs[1]
box = ax.get_position()
h, l = ax.get_legend_handles_labels()
ax.legend(handles=h, loc='upper center',bbox_to_anchor=(0.5, legend_pos), fancybox=True, shadow=True, ncol=ncol, labels=["s=0","s=0.01","s=0.1"])


plt.tight_layout()
#plt.savefig("./figures/figure_1_2_difference_fixation_time_constant_population_size.png", dpi=500)
```
