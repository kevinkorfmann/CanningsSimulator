```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import os

%matplotlib inline
```

```python
sns.set(font_scale = 1.3, style="whitegrid")
```

```python
#files = os.listdir("./simulations_old_floor/")
#files = ["./simulations_old_floor/" + file for file in files if "probability" in file]

files = os.listdir("../simulations/")
files = ["../simulations/" + file for file in files if "probability" in file and "both" in file]

# filter the times
df_time = [pd.read_csv(file) for file in files if "probability" in file]
df_time = pd.concat(df_time).reset_index()

df = df_time[df_time["population_size_model"] == "constant"]
```

```python
df = df[df["population_size"] == 500]

df_viability_01 = df[df["selection_coefficient"] == 0.1].reset_index()
df_viability_001 = df[df["selection_coefficient"] == 0.01].reset_index()

df_fecundity_01 = df[df["selection_coefficient2"] == 0.1].reset_index()
df_fecundity_001 = df[df["selection_coefficient2"] == 0.01].reset_index()
```

```python
colors = "blue"
color = "blue"
linewidth = 2
sns.set_style("whitegrid", {'axes.edgecolor' : '.0001'})

fig, axs = plt.subplots(1, 4, figsize=(19, 5), sharey=True)


ax = axs[0]
sns.lineplot(data=df_viability_001, x="parameter", y="fixation_probability", style="selection_coefficient2", ax=ax, linewidth=linewidth,  color = color).set_title("A) viability selection s=0.01 (N=500)", loc="left")
sns.scatterplot(data=df_viability_001, x="parameter", y="fixation_probability", style="selection_coefficient2", ax=ax, linewidth=linewidth,  color = color, legend=False).set_title("A) viability selection s=0.01 (N=500)", loc="left")
ax.set(xlabel='α', ylabel='fixation probability')
ins = ax.inset_axes([0.29, 0.67, 0.25, 0.25])
sns.lineplot(data=df_viability_001, x="parameter", y="fixation_probability", style="selection_coefficient2", ax=ins, linewidth=linewidth,  color = color, legend=False)
ins.set_yscale("log")
ins.set(ylim=(0.00001, 0.1), xlabel='', ylabel='log(fixation\nprobability)')
ins.patch.set_edgecolor('black')  
ins.tick_params(bottom=True, top=False, left=True, right=True)
ins.patch.set_linewidth('2')
legend_pos=-0.15    
ncol=2
box = ax.get_position()
h, l = ax.get_legend_handles_labels()
leg = ax.legend(handles=h, loc='upper center',bbox_to_anchor=(0.5, legend_pos), fancybox=True, shadow=False, ncol=ncol, title='fecundity selection coefficient')
for line in leg.get_lines():
    line.set_linewidth(2.1) 

    
    
    
ax = axs[1]
sns.lineplot(data=df_viability_01, x="parameter", y="fixation_probability", style="selection_coefficient2", ax=ax, linewidth=linewidth,  color = color).set_title("B) viability selection s=0.1 (N=500)", loc="left")
sns.scatterplot(data=df_viability_01, x="parameter", y="fixation_probability", style="selection_coefficient2", ax=ax, linewidth=linewidth,  color = color, legend=False).set_title("B) viability selection s=0.1 (N=500)", loc="left")
ax.set(xlabel='α', ylabel='fixation probability')
ax.legend(title='fecundity selection coefficient')
ins = ax.inset_axes([0.29, 0.67, 0.25, 0.25])
sns.lineplot(data=df_viability_01, x="parameter", y="fixation_probability", style="selection_coefficient2", ax=ins, linewidth=linewidth,  color = color, legend=False)
ins.set_yscale("log")
ins.set(ylim=(0.00001, 0.1), xlabel='', ylabel='log(fixation\nprobability)')
ins.patch.set_edgecolor('black')  
ins.tick_params(bottom=True, top=False, left=True, right=True)
ins.patch.set_linewidth('2')
legend_pos=-0.15    
ncol=2
box = ax.get_position()
h, l = ax.get_legend_handles_labels()
leg = ax.legend(handles=h, loc='upper center',bbox_to_anchor=(0.5, legend_pos), fancybox=True, shadow=False, ncol=ncol, title='fecundity selection coefficient')
for line in leg.get_lines():
    line.set_linewidth(2.1) 





ax = axs[2]
sns.lineplot(data=df_fecundity_001, x="parameter", y="fixation_probability", style="selection_coefficient", ax=ax, linewidth=linewidth,  color = color).set_title("C) fecundity selection s=0.01 (N=500)", loc="left")
sns.scatterplot(data=df_fecundity_001, x="parameter", y="fixation_probability", style="selection_coefficient", ax=ax, linewidth=linewidth,  color = color, legend=False)
ax.set(xlabel='α', ylabel='fixation probability')
ax.legend(title='fecundity selection coefficient')
ins = ax.inset_axes([0.29, 0.67, 0.25, 0.25])
sns.lineplot(data=df_fecundity_001, x="parameter", y="fixation_probability", style="selection_coefficient", ax=ins, linewidth=linewidth,  color = color, legend=False)
ins.set_yscale("log")
ins.set(ylim=(0.00001, 0.1), xlabel='', ylabel='log(fixation\nprobability)')
ins.patch.set_edgecolor('black')  
ins.tick_params(bottom=True, top=False, left=True, right=True)
ins.patch.set_linewidth('2')
legend_pos=-0.15    
ncol=2
box = ax.get_position()
h, l = ax.get_legend_handles_labels()
leg = ax.legend(handles=h, loc='upper center',bbox_to_anchor=(0.5, legend_pos), fancybox=True, shadow=False, ncol=ncol, title='viability selection coefficient')
for line in leg.get_lines():
    line.set_linewidth(2.1) 

    
    
    
ax = axs[3]
sns.lineplot(data=df_fecundity_01, x="parameter", y="fixation_probability", style="selection_coefficient", ax=ax, linewidth=linewidth,  color = color).set_title("D) fecundity selection s=0.1 (N=500)", loc="left")
sns.scatterplot(data=df_fecundity_01, x="parameter", y="fixation_probability", style="selection_coefficient", ax=ax, linewidth=linewidth,  color = color, legend=False)
ax.set(xlabel='α', ylabel='fixation probability')
ax.legend(title='fecundity selection coefficient')
ins = ax.inset_axes([0.29, 0.67, 0.25, 0.25])
sns.lineplot(data=df_fecundity_01, x="parameter", y="fixation_probability", style="selection_coefficient", ax=ins, linewidth=linewidth,  color = color, legend=False)
ins.set_yscale("log")
ins.set(ylim=(0.00001, 0.1), xlabel='', ylabel='log(fixation\nprobability)')
ins.patch.set_edgecolor('black')  
ins.tick_params(bottom=True, top=False, left=True, right=True)
ins.patch.set_linewidth('2')
legend_pos=-0.15    
ncol=2
box = ax.get_position()
h, l = ax.get_legend_handles_labels()
leg = ax.legend(handles=h, loc='upper center',bbox_to_anchor=(0.5, legend_pos), fancybox=True, shadow=False, ncol=ncol, title='viability selection coefficient')
for line in leg.get_lines():
    line.set_linewidth(2.1) 
    
    
plt.tight_layout()
plt.savefig("./figures/probability_fixation_population_constant_selection_both_500.pdf", dpi=500)
```
