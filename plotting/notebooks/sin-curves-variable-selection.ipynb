```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
import os

%matplotlib inline
```

```python

sns.set(font_scale = 1.4, style="whitegrid")
fig, axs = plt.subplots(4,2, figsize=(18, 12))

ax = axs[0][0]
N = 1
period = 0.06283
start ,stop, step = 0, 500, 1
x = np.arange(start, stop, step)  
y = 0.1 * np.sin(period * x) #(-1)
sns.lineplot(x=x, y=y, ax=ax, color="black").set_title("Fluctuating selection coefficient\n[0.1 sin(0.06283 x)]")
ax.grid(True)
ax.tick_params(bottom=True, top=False, left=True, right=True)
ax.patch.set_edgecolor('black')  
ax.patch.set_linewidth('2') 
ax.set(xlabel='generations', ylabel='selection coefficient')

ax = axs[0][1]
N = 1
period = 0.06283
start ,stop, step = 0, 500, 1
x = np.arange(start, stop, step)  
y = 0.1 * np.sin(period * x + np.pi) #(-1)
sns.lineplot(x=x, y=y, ax=ax, color="black").set_title("Fluctuating selection coefficient\n[(0.1 sin(0.06283 x + $\pi$))]")
ax.grid(True)
ax.tick_params(bottom=True, top=False, left=True, right=True)
ax.patch.set_edgecolor('black')  
ax.patch.set_linewidth('2') 
ax.set(xlabel='generations', ylabel='selection coefficient')


ax = axs[1][0]
N = 1
period = 0.06283
start ,stop, step = 0, 500, 1
x = np.arange(start, stop, step)  
y = 0.01 * np.sin(period * x) #(-1)
sns.lineplot(x=x, y=y, ax=ax, color="black").set_title("Fluctuating selection coefficient\n[0.01 sin(0.06283 x)]")
ax.grid(True)
ax.tick_params(bottom=True, top=False, left=True, right=True)
ax.patch.set_edgecolor('black')  
ax.patch.set_linewidth('2') 
ax.set(xlabel='generations', ylabel='selection coefficient')

ax = axs[1][1]
N = 1
period = 0.06283
start ,stop, step = 0, 500, 1
x = np.arange(start, stop, step)  
y = 0.01 * np.sin(period * x + np.pi) #(-1)
sns.lineplot(x=x, y=y, ax=ax, color="black").set_title("Fluctuating selection coefficient\n[(0.01 sin(0.06283 x + $\pi$))]")
ax.grid(True)
ax.tick_params(bottom=True, top=False, left=True, right=True)
ax.patch.set_edgecolor('black')  
ax.patch.set_linewidth('2') 
ax.set(xlabel='generations', ylabel='selection coefficient')


ax = axs[2][0]
N = 1
period = 0.006283
start ,stop, step = 0, 5000, 1
x = np.arange(start, stop, step)  
y = 0.1 * np.sin(period * x) #(-1)
sns.lineplot(x=x, y=y, ax=ax, color="black").set_title("Fluctuating selection coefficient\n[0.1 sin(0.006283 x)]")
ax.grid(True)
ax.tick_params(bottom=True, top=False, left=True, right=True)
ax.patch.set_edgecolor('black')  
ax.patch.set_linewidth('2') 
ax.set(xlabel='generations', ylabel='selection coefficient')

ax = axs[2][1]
N = 1
period = 0.006283
start ,stop, step = 0, 5000, 1
x = np.arange(start, stop, step)  
y = 0.1 * np.sin(period * x + np.pi) #(-1)
sns.lineplot(x=x, y=y, ax=ax, color="black").set_title("Fluctuating selection coefficient\n[(0.1 sin(0.006283 x + $\pi$))]")
ax.grid(True)
ax.tick_params(bottom=True, top=False, left=True, right=True)
ax.patch.set_edgecolor('black')  
ax.patch.set_linewidth('2') 
ax.set(xlabel='generations', ylabel='selection coefficient')


ax = axs[3][0]
N = 1
period = 0.006283
start ,stop, step = 0, 5000, 1
x = np.arange(start, stop, step)  
y = 0.01 * np.sin(period * x) #(-1)
sns.lineplot(x=x, y=y, ax=ax, color="black").set_title("Fluctuating selection coefficient\n[0.01 sin(0.006283 x)]")
ax.grid(True)
ax.tick_params(bottom=True, top=False, left=True, right=True)
ax.patch.set_edgecolor('black')  
ax.patch.set_linewidth('2') 
ax.set(xlabel='generations', ylabel='selection coefficient')

ax = axs[3][1]
N = 1
period = 0.006283
start ,stop, step = 0, 5000, 1
x = np.arange(start, stop, step)  
y = 0.01 * np.sin(period * x + np.pi) #(-1)
sns.lineplot(x=x, y=y, ax=ax, color="black").set_title("Fluctuating selection coefficient\n[(0.01 sin(0.006283 x + $\pi$))]")
ax.grid(True)
ax.tick_params(bottom=True, top=False, left=True, right=True)
ax.patch.set_edgecolor('black')  
ax.patch.set_linewidth('2') 
ax.set(xlabel='generations', ylabel='selection coefficient')


plt.tight_layout()
plt.savefig("./figures/fluctuating_selection_coefficient.pdf", dpi=500)

```
