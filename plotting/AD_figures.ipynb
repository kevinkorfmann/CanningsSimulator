```python
import os
from collections import defaultdict

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np

# ======= SETTINGS =======
sns.set(font_scale=1.3, style="whitegrid")
sns.set_style("whitegrid", {'axes.edgecolor': '.0001'})

PERIOD_SLOW = "0.006283"   # slow
PERIOD_FAST = "0.06283"    # fast
SHIFT_A = "shifted_0.0"    # phase A (initially advantageous)
SHIFT_B = "shifted_3.1"    # phase B (initially deleterious)
DATA_DIR = "../src/simulations/"
POP_N = 1000
LINEWIDTH = 2

# canonical linestyle patterns
CANON_DASHES = {
    "0.0": (),            # solid
    "0.01": (6, 3),       # dashed
    "0.1": (1, 3),        # dotted
}
CANON_S_ORDER = ["0.0", "0.01", "0.1"]
PHASE_ORDER = [SHIFT_A, SHIFT_B, "neutral"]

# palettes per side (phase → color)
PALETTE_SLOW = {
    SHIFT_A: "#1f77b4",   # dark blue
    SHIFT_B: "#9ecae1",   # light blue
    "neutral": "0.6",     # gray
}
PALETTE_FAST = {
    SHIFT_A: "#d62728",   # dark red
    SHIFT_B: "#f28e8e",   # light red
    "neutral": "0.6",     # gray
}

# ======= HELPERS =======
def load_csvs(pattern_func):
    paths = [os.path.join(DATA_DIR, f) for f in os.listdir(DATA_DIR) if pattern_func(f)]
    dfs = []
    for p in paths:
        try:
            dfs.append(pd.read_csv(p))
        except Exception as e:
            print(f"[WARN] Could not read {p}: {e}")
    if not dfs:
        return pd.DataFrame()
    return pd.concat(dfs, ignore_index=True, sort=False)

def add_phase_period(df, phase, period):
    if df.empty:
        return df
    out = df.copy()
    out["phase"] = phase
    out["period"] = period
    return out

def only_constant(df):
    return df[df["population_size_model"] == "constant"] if not df.empty else df

def report_missing(df, label, y_col):
    if df.empty:
        print(f("[WARN] {label}: no data"))
        return
    have = df.groupby(
        ["phase", "period", "selection_type", "parameter",
         "selection_coefficient", "population_size"],
        dropna=False
    )[y_col].size()
    print(f"[INFO] {label}: {len(have)} groups; head:")
    print(have.reset_index().head(10).to_string(index=False))

def build_panel(dfA, dfB, neu_df, sel_type):
    parts = []
    for d in (dfA, dfB):
        if not d.empty:
            parts.append(d[d["selection_type"] == sel_type])
    if not neu_df.empty:
        parts.append(neu_df[neu_df["selection_type"] == sel_type])
    if not parts:
        return pd.DataFrame()
    out = pd.concat(parts, ignore_index=True, sort=False)
    return out

def pick_y_column(df):
    for c in ("fixation_probability", "fixation_time"):
        if c in df.columns:
            return c
    return None

def normalize_selection_coefficients(df):
    """
    Convert selection_coefficient to canonical magnitude strings:
    - If encoded as multiplier (e.g. 1.01, 0.99), use |value - 1|.
    - If encoded as raw s (e.g. 0.01, -0.01), use |value|.
    Then snap to {"0.0","0.01","0.1"} when numerically close.
    """
    if df.empty or "selection_coefficient" not in df:
        return df
    out = df.copy()
    def to_mag_str(x):
        try:
            v = float(x)
        except Exception:
            return str(x)
        s_mag = abs(v - 1.0) if v > 0.5 else abs(v)
        # snap to canonical bins
        if np.isclose(s_mag, 0.0, atol=1e-6):
            return "0.0"
        if np.isclose(s_mag, 0.01, atol=5e-4):
            return "0.01"
        if np.isclose(s_mag, 0.1, atol=5e-3):
            return "0.1"
        # fallback: round to 3 decimals
        return f"{s_mag:.3f}"
    out["selection_coefficient"] = out["selection_coefficient"].map(to_mag_str)
    return out

def plot_panel(ax, df, title, palette, y_col):
    if df.empty:
        ax.set_title(title + " (no data)", loc="left")
        ax.set(xlabel='α', ylabel='Fixation probability')
        ax.grid(True)
        ax.patch.set_edgecolor('black')
        ax.patch.set_linewidth(1.5)
        return

    df = normalize_selection_coefficients(df)

    # determine which s-levels actually exist in this panel
    present_s = [s for s in CANON_S_ORDER if s in set(df["selection_coefficient"])]
    # include any non-canonical leftovers (rare) to avoid KeyError, putting them at the end
    leftovers = [s for s in sorted(df["selection_coefficient"].unique()) if s not in present_s]
    s_order = present_s + leftovers

    # build a dashes dict that covers all present keys
    dashes = {**{k: CANON_DASHES[k] for k in present_s}}
    # give a distinct fallback pattern for leftovers
    for i, k in enumerate(leftovers):
        dashes[k] = (4, 2, 1, 2) if i % 2 == 0 else (8, 2)

    sns.lineplot(
        data=df, x="parameter", y=y_col,
        hue="phase", hue_order=PHASE_ORDER, palette=palette,
        style="selection_coefficient", style_order=s_order,
        dashes=dashes, errorbar=None, ax=ax, linewidth=LINEWIDTH, legend=False
    )
    sns.scatterplot(
        data=df, x="parameter", y=y_col,
        hue="phase", hue_order=PHASE_ORDER, palette=palette,
        style="selection_coefficient", style_order=s_order,
        markers=True, ax=ax, legend=False, s=20
    )

    ax.set_title(title, loc="left")
    ax.set(xlabel='α', ylabel='Fixation probability' if y_col == "fixation_probability" else 'Fixation time')
    ax.grid(True)
    ax.patch.set_edgecolor('black')
    ax.patch.set_linewidth(1.5)

# ======= LOAD DATA =======
# Neutral (independent of period) → map to sinus labels + phase="neutral"
df_neutral = load_csvs(lambda f: ("time" in f and "constant_selection_0.0_" in f and "adapted" not in f))
df_neutral = only_constant(df_neutral)

viab_neu = df_neutral[df_neutral["selection_type"] == "viability_constant"].copy()
fec_neu  = df_neutral[df_neutral["selection_type"] == "fecundity_constant"].copy()
viab_neu["selection_type"] = "viability_sinus"
fec_neu["selection_type"]  = "fecundity_sinus"
for d in (viab_neu, fec_neu):
    d["phase"] = "neutral"
    d["period"] = "neutral"

# SLOW base + overlay
slow_A = load_csvs(lambda f: ("time" in f and SHIFT_A in f and "period" in f and PERIOD_SLOW in f))
slow_B = load_csvs(lambda f: ("time" in f and SHIFT_B in f and "period" in f and PERIOD_SLOW in f))
slow_A = add_phase_period(only_constant(slow_A), SHIFT_A, PERIOD_SLOW)
slow_B = add_phase_period(only_constant(slow_B), SHIFT_B, PERIOD_SLOW)

# FAST base + overlay
fast_A = load_csvs(lambda f: ("time" in f and SHIFT_A in f and "period" in f and PERIOD_FAST in f))
fast_B = load_csvs(lambda f: ("time" in f and SHIFT_B in f and "period" in f and PERIOD_FAST in f))
fast_A = add_phase_period(only_constant(fast_A), SHIFT_A, PERIOD_FAST)
fast_B = add_phase_period(only_constant(fast_B), SHIFT_B, PERIOD_FAST)

# Determine y column
candidates = [slow_A, slow_B, fast_A, fast_B, viab_neu, fec_neu]
y_col = None
for cdf in candidates:
    y_col = y_col or pick_y_column(cdf)
if y_col is None:
    raise RuntimeError("Could not find 'fixation_probability' or 'fixation_time' in any CSVs.")

# Build panels and filter by N
viab_slow = build_panel(slow_A, slow_B, viab_neu, "viability_sinus")
fec_slow  = build_panel(slow_A, slow_B, fec_neu,  "fecundity_sinus")
viab_fast = build_panel(fast_A, fast_B, viab_neu, "viability_sinus")
fec_fast  = build_panel(fast_A, fast_B, fec_neu,  "fecundity_sinus")

viab_slow = viab_slow[viab_slow["population_size"] == POP_N]
fec_slow  = fec_slow [fec_slow ["population_size"] == POP_N]
viab_fast = viab_fast[viab_fast["population_size"] == POP_N]
fec_fast  = fec_fast [fec_fast ["population_size"] == POP_N]

# Normalize s encodings now (also helps the missing-data report)
viab_slow = normalize_selection_coefficients(viab_slow)
fec_slow  = normalize_selection_coefficients(fec_slow)
viab_fast = normalize_selection_coefficients(viab_fast)
fec_fast  = normalize_selection_coefficients(fec_fast)

# Quick console summaries
report_missing(viab_slow, "viab_slow", y_col)
report_missing(fec_slow,  "fec_slow",  y_col)
report_missing(viab_fast, "viab_fast", y_col)
report_missing(fec_fast,  "fec_fast",  y_col)

# ======= PLOT =======
fig, axs = plt.subplots(2, 2, figsize=(18, 12), sharey=True)

# (a) slow viability (top-left)
plot_panel(
    axs[0][0],
    viab_slow,
    f"(a) Slow fluctuating viability selection ($N={POP_N}$)",
    PALETTE_SLOW,
    y_col
)

# (c) slow fecundity (bottom-left)
plot_panel(
    axs[1][0],
    fec_slow,
    f"(c) Slow fluctuating fecundity selection ($N={POP_N}$)",
    PALETTE_SLOW,
    y_col
)

# (b) fast viability (top-right)
plot_panel(
    axs[0][1],
    viab_fast,
    f"(b) Fast fluctuating viability selection ($N={POP_N}$)",
    PALETTE_FAST,
    y_col
)

# (d) fast fecundity (bottom-right)
plot_panel(
    axs[1][1],
    fec_fast,
    f"(d) Fast fluctuating fecundity selection ($N={POP_N}$)",
    PALETTE_FAST,
    y_col
)

# Global legend for s (linestyle only; canonical 0, 0.01, 0.1)
handles = [
    Line2D([0], [0], color='0.2', linewidth=LINEWIDTH, linestyle='-',              label='s = 0'),
    Line2D([0], [0], color='0.2', linewidth=LINEWIDTH, linestyle=(0, (6, 3)),     label='s = 0.01'),
    Line2D([0], [0], color='0.2', linewidth=LINEWIDTH, linestyle=(0, (1, 3)),     label='s = 0.1'),
]
fig.legend(handles=handles, loc='lower center', ncol=3, frameon=True)

plt.tight_layout(rect=(0, 0.06, 1, 1))
plt.savefig("./figures/figure_6_probabilities.pdf", dpi=500)
plt.show()
```

```python
import os
from collections import defaultdict

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import numpy as np

# ======= SETTINGS =======
sns.set(font_scale=1.3, style="whitegrid")
sns.set_style("whitegrid", {'axes.edgecolor': '.0001'})

PERIOD_SLOW = "0.006283"   # slow
PERIOD_FAST = "0.06283"    # fast
SHIFT_A = "shifted_0.0"    # phase A (initially advantageous)
SHIFT_B = "shifted_3.1"    # phase B (initially deleterious)
DATA_DIR = "../src/simulations/"
POP_N = 1000
LINEWIDTH = 2

# canonical linestyle patterns
CANON_DASHES = {
    "0.0": (),            # solid
    "0.01": (6, 3),       # dashed
    "0.1": (1, 3),        # dotted
}
CANON_S_ORDER = ["0.0", "0.01", "0.1"]
PHASE_ORDER = [SHIFT_A, SHIFT_B, "neutral"]

# palettes per side (phase → color)
PALETTE_SLOW = {
    SHIFT_A: "#1f77b4",   # dark blue
    SHIFT_B: "#9ecae1",   # light blue
    "neutral": "0.6",     # gray
}
PALETTE_FAST = {
    SHIFT_A: "#d62728",   # dark red
    SHIFT_B: "#f28e8e",   # light red
    "neutral": "0.6",     # gray
}

# ======= HELPERS =======
def load_csvs(pattern_func):
    paths = [os.path.join(DATA_DIR, f) for f in os.listdir(DATA_DIR) if pattern_func(f)]
    dfs = []
    for p in paths:
        try:
            dfs.append(pd.read_csv(p))
        except Exception as e:
            print(f"[WARN] Could not read {p}: {e}")
    if not dfs:
        return pd.DataFrame()
    return pd.concat(dfs, ignore_index=True, sort=False)

def add_phase_period(df, phase, period):
    if df.empty:
        return df
    out = df.copy()
    out["phase"] = phase
    out["period"] = period
    return out

def only_constant(df):
    return df[df["population_size_model"] == "constant"] if not df.empty else df

def report_missing(df, label, y_col):
    if df.empty:
        print(f("[WARN] {label}: no data"))
        return
    have = df.groupby(
        ["phase", "period", "selection_type", "parameter",
         "selection_coefficient", "population_size"],
        dropna=False
    )[y_col].size()
    print(f"[INFO] {label}: {len(have)} groups; head:")
    print(have.reset_index().head(10).to_string(index=False))

def build_panel(dfA, dfB, neu_df, sel_type):
    parts = []
    for d in (dfA, dfB):
        if not d.empty:
            parts.append(d[d["selection_type"] == sel_type])
    if not neu_df.empty:
        parts.append(neu_df[neu_df["selection_type"] == sel_type])
    if not parts:
        return pd.DataFrame()
    out = pd.concat(parts, ignore_index=True, sort=False)
    return out

def pick_y_column(df):
    for c in ("fixation_probability", "fixation_time"):
        if c in df.columns:
            return c
    return None

def normalize_selection_coefficients(df):
    """
    Convert selection_coefficient to canonical magnitude strings:
    - If encoded as multiplier (e.g. 1.01, 0.99), use |value - 1|.
    - If encoded as raw s (e.g. 0.01, -0.01), use |value|.
    Then snap to {"0.0","0.01","0.1"} when numerically close.
    """
    if df.empty or "selection_coefficient" not in df:
        return df
    out = df.copy()
    def to_mag_str(x):
        try:
            v = float(x)
        except Exception:
            return str(x)
        s_mag = abs(v - 1.0) if v > 0.5 else abs(v)
        # snap to canonical bins
        if np.isclose(s_mag, 0.0, atol=1e-6):
            return "0.0"
        if np.isclose(s_mag, 0.01, atol=5e-4):
            return "0.01"
        if np.isclose(s_mag, 0.1, atol=5e-3):
            return "0.1"
        # fallback: round to 3 decimals
        return f"{s_mag:.3f}"
    out["selection_coefficient"] = out["selection_coefficient"].map(to_mag_str)
    return out

def plot_panel(ax, df, title, palette, y_col):
    if df.empty:
        ax.set_title(title + " (no data)", loc="left")
        ax.set(xlabel='α', ylabel='Fixation probability')
        ax.grid(True)
        ax.patch.set_edgecolor('black')
        ax.patch.set_linewidth(1.5)
        return

    df = normalize_selection_coefficients(df)

    # determine which s-levels actually exist in this panel
    present_s = [s for s in CANON_S_ORDER if s in set(df["selection_coefficient"])]
    # include any non-canonical leftovers (rare) to avoid KeyError, putting them at the end
    leftovers = [s for s in sorted(df["selection_coefficient"].unique()) if s not in present_s]
    s_order = present_s + leftovers

    # build a dashes dict that covers all present keys
    dashes = {**{k: CANON_DASHES[k] for k in present_s}}
    # give a distinct fallback pattern for leftovers
    for i, k in enumerate(leftovers):
        dashes[k] = (4, 2, 1, 2) if i % 2 == 0 else (8, 2)

    sns.lineplot(
        data=df, x="parameter", y=y_col,
        hue="phase", hue_order=PHASE_ORDER, palette=palette,
        style="selection_coefficient", style_order=s_order,
        dashes=dashes, errorbar=None, ax=ax, linewidth=LINEWIDTH, legend=False
    )
    sns.scatterplot(
        data=df, x="parameter", y=y_col,
        hue="phase", hue_order=PHASE_ORDER, palette=palette,
        style="selection_coefficient", style_order=s_order,
        markers=True, ax=ax, legend=False, s=20
    )

    ax.set_title(title, loc="left")
    ax.set(xlabel='α', ylabel='Fixation probability' if y_col == "fixation_probability" else 'Fixation time')
    ax.grid(True)
    ax.patch.set_edgecolor('black')
    ax.patch.set_linewidth(1.5)

# ======= LOAD DATA =======
# Neutral (independent of period) → map to sinus labels + phase="neutral"
df_neutral = load_csvs(lambda f: ("probability" in f and "constant_selection_0.0_" in f and "adapted" not in f))
df_neutral = only_constant(df_neutral)

viab_neu = df_neutral[df_neutral["selection_type"] == "viability_constant"].copy()
fec_neu  = df_neutral[df_neutral["selection_type"] == "fecundity_constant"].copy()
viab_neu["selection_type"] = "viability_sinus"
fec_neu["selection_type"]  = "fecundity_sinus"
for d in (viab_neu, fec_neu):
    d["phase"] = "neutral"
    d["period"] = "neutral"

# SLOW base + overlay
slow_A = load_csvs(lambda f: ("probability" in f and SHIFT_A in f and "period" in f and PERIOD_SLOW in f))
slow_B = load_csvs(lambda f: ("probability" in f and SHIFT_B in f and "period" in f and PERIOD_SLOW in f))
slow_A = add_phase_period(only_constant(slow_A), SHIFT_A, PERIOD_SLOW)
slow_B = add_phase_period(only_constant(slow_B), SHIFT_B, PERIOD_SLOW)

# FAST base + overlay
fast_A = load_csvs(lambda f: ("probability" in f and SHIFT_A in f and "period" in f and PERIOD_FAST in f))
fast_B = load_csvs(lambda f: ("probability" in f and SHIFT_B in f and "period" in f and PERIOD_FAST in f))
fast_A = add_phase_period(only_constant(fast_A), SHIFT_A, PERIOD_FAST)
fast_B = add_phase_period(only_constant(fast_B), SHIFT_B, PERIOD_FAST)

# Determine y column
candidates = [slow_A, slow_B, fast_A, fast_B, viab_neu, fec_neu]
y_col = None
for cdf in candidates:
    y_col = y_col or pick_y_column(cdf)
if y_col is None:
    raise RuntimeError("Could not find 'fixation_probability' or 'fixation_time' in any CSVs.")

# Build panels and filter by N
viab_slow = build_panel(slow_A, slow_B, viab_neu, "viability_sinus")
fec_slow  = build_panel(slow_A, slow_B, fec_neu,  "fecundity_sinus")
viab_fast = build_panel(fast_A, fast_B, viab_neu, "viability_sinus")
fec_fast  = build_panel(fast_A, fast_B, fec_neu,  "fecundity_sinus")

viab_slow = viab_slow[viab_slow["population_size"] == POP_N]
fec_slow  = fec_slow [fec_slow ["population_size"] == POP_N]
viab_fast = viab_fast[viab_fast["population_size"] == POP_N]
fec_fast  = fec_fast [fec_fast ["population_size"] == POP_N]

# Normalize s encodings now (also helps the missing-data report)
viab_slow = normalize_selection_coefficients(viab_slow)
fec_slow  = normalize_selection_coefficients(fec_slow)
viab_fast = normalize_selection_coefficients(viab_fast)
fec_fast  = normalize_selection_coefficients(fec_fast)

# Quick console summaries
report_missing(viab_slow, "viab_slow", y_col)
report_missing(fec_slow,  "fec_slow",  y_col)
report_missing(viab_fast, "viab_fast", y_col)
report_missing(fec_fast,  "fec_fast",  y_col)

# ======= PLOT =======
fig, axs = plt.subplots(2, 2, figsize=(18, 12), sharey=True)

# (a) slow viability (top-left)
plot_panel(
    axs[0][0],
    viab_slow,
    f"(a) Slow fluctuating viability selection ($N={POP_N}$)",
    PALETTE_SLOW,
    y_col
)

# (c) slow fecundity (bottom-left)
plot_panel(
    axs[1][0],
    fec_slow,
    f"(c) Slow fluctuating fecundity selection ($N={POP_N}$)",
    PALETTE_SLOW,
    y_col
)

# (b) fast viability (top-right)
plot_panel(
    axs[0][1],
    viab_fast,
    f"(b) Fast fluctuating viability selection ($N={POP_N}$)",
    PALETTE_FAST,
    y_col
)

# (d) fast fecundity (bottom-right)
plot_panel(
    axs[1][1],
    fec_fast,
    f"(d) Fast fluctuating fecundity selection ($N={POP_N}$)",
    PALETTE_FAST,
    y_col
)

# Global legend for s (linestyle only; canonical 0, 0.01, 0.1)
handles = [
    Line2D([0], [0], color='0.2', linewidth=LINEWIDTH, linestyle='-',              label='s = 0'),
    Line2D([0], [0], color='0.2', linewidth=LINEWIDTH, linestyle=(0, (6, 3)),     label='s = 0.01'),
    Line2D([0], [0], color='0.2', linewidth=LINEWIDTH, linestyle=(0, (1, 3)),     label='s = 0.1'),
]
fig.legend(handles=handles, loc='lower center', ncol=3, frameon=True)

plt.tight_layout(rect=(0, 0.06, 1, 1))
# plt.savefig("./figures/figure_6_probabilities.pdf", dpi=500)
plt.show()
```

```python
import os
import pandas as pd

# ---------- CONFIG ----------
DATA_DIR     = "../src/simulations/"
POP_SIZE     = 1000
PERIOD_SLOW  = "0.006283"   # slow
PERIOD_FAST  = "0.06283"    # fast
SHIFT_A      = "shifted_0.0"
SHIFT_B      = "shifted_3.1"
SHIFT_PI     = "3.141592653"   # --shiftsinusby value for SHIFT_B files

# Which metrics do you want commands for? e.g., ["time"] or ["probability"] or both.
METRICS      = ["time"]      # change to ["time","probability"] if you want both

# Selection types to inspect
SINUS_TYPES   = ["viability_sinus", "fecundity_sinus"]

# Sim settings per metric
NB_SIMS       = {"time": 5000, "probability": 500000}
OUTPUT_NAME   = {"time": "fixation_time", "probability": "fixation_probability"}

# ---------- HELPERS ----------
def list_files():
    return [os.path.join(DATA_DIR, f) for f in os.listdir(DATA_DIR)]

def read_many(paths):
    if not paths:
        return pd.DataFrame()
    return pd.concat([pd.read_csv(p) for p in paths], ignore_index=True)

def files_period(metric, period, shift):
    return [p for p in list_files() if (metric in p and "period" in p and period in p and shift in p)]

def files_neutral(metric):
    # neutral files: contain metric, no "period", include "constant_selection_0.0_", exclude "adapted"
    return [p for p in list_files()
            if (metric in p and "period" not in p and "constant_selection_0.0_" in p and "adapted" not in p)]

def load_block(metric, period, shift):
    """Return (df_sinus, df_neutral) filtered to population_size_model=='constant'."""
    dfp = read_many(files_period(metric, period, shift))
    if not dfp.empty:
        dfp = dfp[dfp["population_size_model"] == "constant"].reset_index(drop=True)
    dfn = read_many(files_neutral(metric))
    if not dfn.empty:
        dfn = dfn[dfn["population_size_model"] == "constant"].reset_index(drop=True)
    return dfp, dfn

def subset(df, sel_type, pop_size):
    if df is None or df.empty:
        return pd.DataFrame()
    out = df[df["selection_type"] == sel_type]
    if "population_size" in out.columns:
        out = out[out["population_size"] == pop_size]
    return out.reset_index(drop=True)

def param_set(df):
    # Return (set_of_selection_coeffs, dict sc -> set(parameters))
    if df is None or df.empty:
        return set(), {}
    scs = set(df["selection_coefficient"].dropna().tolist()) if "selection_coefficient" in df.columns else set()
    params_by_sc = {s: set(df.loc[df["selection_coefficient"] == s, "parameter"].dropna().tolist()) for s in scs}
    return scs, params_by_sc

def union_params(param_maps):
    """param_maps: list of dicts sc->set(params)."""
    union_map = {}
    for m in param_maps:
        if not isinstance(m, dict):
            continue
        for s, ps in m.items():
            union_map.setdefault(s, set()).update(ps)
    all_scs = set(union_map.keys())
    return all_scs, union_map

def missing_report_for(sel_type, blocks):
    # Build per-block param maps (sinus only)
    by_block = {}
    for name, (dfp, dfn) in blocks.items():
        df_sel = subset(dfp, sel_type, POP_SIZE)
        scs, pmap = param_set(df_sel)
        by_block[name] = {"scs": scs, "pmap": pmap}

    # expected grid = union across all panels
    pmap_list = [info["pmap"] for info in by_block.values()]
    exp_scs, exp_pmap = union_params(pmap_list)

    rows = []
    for name, info in by_block.items():
        pmap_present = info["pmap"]
        for s in sorted(exp_scs):
            exp_params = sorted(exp_pmap.get(s, set()))
            have_params = sorted(pmap_present.get(s, set())) if s in pmap_present else []
            miss = sorted(set(exp_params) - set(have_params))
            if miss:
                rows.append({
                    "panel": name,
                    "selection_type": sel_type,
                    "selection_coefficient": s,
                    "missing_parameters": miss,
                })
    return pd.DataFrame(rows)

def dupes_report(df_blocks, sel_type):
    rows = []
    for name, (dfp, _) in df_blocks.items():
        d = subset(dfp, sel_type, POP_SIZE)
        if d.empty: 
            continue
        grp = d.groupby(["selection_coefficient", "parameter"]).size().reset_index(name="count")
        dupes = grp[grp["count"] > 1]
        if not dupes.empty:
            for _, r in dupes.iterrows():
                rows.append({
                    "panel": name,
                    "selection_type": sel_type,
                    "selection_coefficient": r["selection_coefficient"],
                    "parameter": r["parameter"],
                    "duplicate_count": int(r["count"]),
                })
    return pd.DataFrame(rows)

# Map our panels to (period, shiftflag)
PANEL_MAP = {
    "slow_shift0_top": (PERIOD_SLOW, None),
    "slow_shift3_top": (PERIOD_SLOW, SHIFT_PI),
    "fast_shift0_top": (PERIOD_FAST, None),
    "fast_shift3_top": (PERIOD_FAST, SHIFT_PI),
}

def to_sim_s(s_coef):
    """Convert CSV selection_coefficient to simulator -s.
    If value is additive (< 0.5), return 1 + s; if already multiplicative (>= 0.5), pass through."""
    s = float(s_coef)
    return s if s >= 0.5 else (1.0 + s)

def make_command(sel_type, s_coef, param, metric, period, shift_pi, pop_size=POP_SIZE):
    s_cli = to_sim_s(s_coef)   # <- use the auto-detector
    nb    = NB_SIMS[metric]
    out   = OUTPUT_NAME[metric]
    base = [
        "julia", "simulator.jl",
        "-N", str(pop_size),
        "-t", sel_type,
        "-s", f"{s_cli:.6g}",
        "-m", "Schweinsberg",
        "-p", f"{float(param):.6g}",
        "-c", "constant",
        "-o", out,
        "--type", metric,
        "--nb_simulations", str(nb),
        "--selection_period", period,
    ]
    if shift_pi is not None:
        base += ["--shiftsinusby", shift_pi]
    return " ".join(base) + " &"

# ---------- LOAD PANELS + FIND MISSING ----------
commands = []
for metric in METRICS:
    blocks = {
        "slow_shift0_top": load_block(metric, PERIOD_SLOW, SHIFT_A),
        "slow_shift3_top": load_block(metric, PERIOD_SLOW, SHIFT_B),
        "fast_shift0_top": load_block(metric, PERIOD_FAST, SHIFT_A),
        "fast_shift3_top": load_block(metric, PERIOD_FAST, SHIFT_B),
    }

    reports = []
    duplicates = []
    for sel_type in SINUS_TYPES:
        reports.append(missing_report_for(sel_type, blocks))
        duplicates.append(dupes_report(blocks, sel_type))
    missing_df = pd.concat(reports, ignore_index=True) if reports else pd.DataFrame()
    dupes_df   = pd.concat(duplicates, ignore_index=True) if duplicates else pd.DataFrame()

    print(f"\n=== METRIC: {metric} | POP_SIZE: {POP_SIZE} ===")
    if missing_df.empty:
        print("✅ No missing (selection_coefficient, parameter) points across panels.")
    else:
        print("❌ Missing datapoints found. Generating commands...")
        # Expand rows into explicit commands
        for _, row in missing_df.iterrows():
            panel = row["panel"]
            sel_type = row["selection_type"]
            s_coef = float(row["selection_coefficient"])
            period, shift_pi = PANEL_MAP[panel]
            for param in row["missing_parameters"]:
                cmd = make_command(sel_type, s_coef, param, metric, period, shift_pi, POP_SIZE)
                commands.append((metric, panel, sel_type, s_coef, param, cmd))

    if not dupes_df.empty:
        print("\n⚠️ Duplicates detected (same selection_coefficient & parameter present multiple times in a panel):")
        print(dupes_df.sort_values(["selection_type", "panel", "selection_coefficient", "parameter"]))

# ---------- PRINT COMMANDS NEEDED ----------
if commands:
    print("\n### Commands to run to fill the missing points ###")
    # sort nicely: metric -> panel order -> sel_type -> s -> param
    panel_order = list(PANEL_MAP.keys())
    commands.sort(key=lambda x: (x[0], panel_order.index(x[1]), x[2], x[3], float(x[4])))
    for _, panel, sel_type, s_coef, param, cmd in commands:
        print(cmd)
else:
    print("\nNo commands needed 🎉")
```
